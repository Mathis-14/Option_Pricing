# Complete Summary of the Greeks (Vanilla Options)

The "Greeks" measure the sensitivity of an option's price to its various risk factors. They allow traders and risk managers to understand their portfolio's behavior and implement hedging strategies.

---

## 1. Primary Greeks

### **Delta ($\Delta$)**
* **Definition**: The ratio of the change in the option's price to the change in the underlying asset's price.
* **Interpretation**: Represents the slope of the option's value curve and the number of units of the underlying asset to hold to create a risk-free portfolio (Delta-neutral).
* **Formulas (Black-Scholes Model)**:
    * **Call**: $\Delta = N(d_1)$ (always positive, between 0 and 1).
    * **Put**: $\Delta = N(d_1) - 1$ (always negative, between -1 and 0).
* **Hedging**: The position must be adjusted periodically (dynamic hedging) because Delta changes as the underlying price moves.

### **Gamma ($\Gamma$)**
* **Definition**: The rate of change of Delta with respect to the underlying price (the second derivative of the price).
* **Formula**: $\Gamma = \frac{N'(d_1)}{S \sigma \sqrt{\tau}}$
* **Insights**: Represents the convexity of the payoff. It is at its maximum when the option is "At-The-Money" (ATM) and increases as the option approaches maturity.

### **Theta ($\Theta$)**
* **Definition**: The sensitivity of the portfolio's value to the passage of time ("time decay").
* **Formulas**:
    * **Call**: $\Theta = -\frac{\sigma S N'(d_1)}{2 \sqrt{\tau}} - r K e^{-r\tau} N(d_2)$
    * **Put**: $\Theta = -\frac{\sigma S N'(d_1)}{2 \sqrt{\tau}} + r K e^{-r\tau} N(-d_2)$
* **Insights**: Theta is generally negative for the option buyer (loss of time value).

### **Vega ($\nu$ or $Vega$)**
* **Definition**: The sensitivity of the portfolio's value to a change in implied volatility.
* **Formula**: $Vega = S N'(d_1) \sqrt{\tau}$
* **Insights**: Vega is the same for both a call and a put. It is higher for long-dated options.

### **Rho ($\rho$)**
* **Definition**: The sensitivity of the portfolio's value to a change in the risk-free interest rate.
* **Formulas**:
    * **Call**: $\rho = K \tau e^{-r\tau} N(d_2)$
    * **Put**: $\rho = -K \tau e^{-r\tau} N(-d_2)$

---

## 2. Higher-Order Convexities
Beyond the first order, traders monitor more complex risks:
* **Vanna**: $\frac{\partial^2 P}{\partial \bar{\sigma} \partial S}$ — Sensitivity of Delta to volatility.
* **Voma**: $\frac{\partial^2 P}{\partial \bar{\sigma}^2}$ — Sensitivity of Vega to volatility.

---

## 3. Key Relationships and Volatility Trading

### **Black-Scholes Equation and P&L**
The P&L of a delta-hedged position over a small time step $dt$ is given by:
$$dP\&L \approx \frac{1}{2} \Gamma S^2 (\sigma_{real}^2 - \sigma_{imp}^2) dt$$
* **Profit**: If realized volatility ($\sigma_{real}$) is higher than the initial implied volatility ($\sigma_{imp}$), the long-gamma position generates a profit.

### **Gamma-Theta Arbitrage**
For a delta-neutral portfolio with interest rates close to zero:
$$\Theta \approx -\frac{1}{2} \sigma^2 S^2 \Gamma$$
* A high positive Theta often implies a high negative Gamma, and vice versa.

### **Vega-Gamma Relationship**
In the Black-Scholes model:
$$Vega = \Gamma S^2 \sigma \tau$$
* Short-dated options are used to trade **Gamma**, while long-dated options are used to trade **Vega**.

## 4. Summary for Underlyings with Dividends ($q$)
| Greek | Call Option | Put Option |
| :--- | :--- | :--- |
| **Delta** | $e^{-q\tau} N(d_1)$ | $e^{-q\tau} (N(d_1) - 1)$ |
| **Gamma** | $\frac{N'(d_1) e^{-q\tau}}{S \sigma \sqrt{\tau}}$ | $\frac{N'(d_1) e^{-q\tau}}{S \sigma \sqrt{\tau}}$ |
| **Vega** | $S \sqrt{\tau} N'(d_1) e^{-q\tau}$ | $S \sqrt{\tau} N'(d_1) e^{-q\tau}$ |
| **Rho** | $K \tau e^{-r\tau} N(d_2)$ | $-K \tau e^{-r\tau} N(-d_2)$ |
