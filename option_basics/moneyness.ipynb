### Quick map to keep in mind on a chart where OTM/ITM is for call/puts 

```python
import matplotlib.pyplot as plt
import numpy as np

# Setup
K = 100
S_min, S_max = 50, 150
S = np.linspace(S_min, S_max, 100)

fig, ax = plt.subplots(figsize=(10, 4))

# 1. Draw the Barrier (The Strike)
ax.axvline(K, color='black', linestyle='--', lw=2)
ax.text(K, -0.1, f'Strike (K={K})', ha='center', fontweight='bold')

# 2. Represent the CALL Regions
ax.axvspan(S_min, K, color='red', alpha=0.1, label='Call OTM / Put ITM')
ax.axvspan(K, S_max, color='green', alpha=0.1, label='Call ITM / Put OTM')

# 3. Add Labels for the Mind
ax.text((S_min + K)/2, 0.5, "CALL: OTM\n(Worthless)", ha='center', color='darkred', fontsize=12)
ax.text((S_min + K)/2, 0.2, "PUT: ITM\n(In the Money)", ha='center', color='darkgreen', fontsize=10)

ax.text((S_max + K)/2, 0.5, "CALL: ITM\n(In the Money)", ha='center', color='darkgreen', fontsize=12)
ax.text((S_max + K)/2, 0.2, "PUT: OTM\n(Worthless)", ha='center', color='darkred', fontsize=10)

# Formatting
ax.set_xlim(S_min, S_max)
ax.set_ylim(0, 1)
ax.set_yticks([])
ax.set_xlabel("Spot Price (S)", fontsize=12)
ax.set_title("The Trader's Mental Map: Moneyness Zones", fontsize=14)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

# Careful ! it is the opposite with strike (K) in x axis !!

```python
# Setup
S = 100  # Fixed Spot Price
K_min, K_max = 50, 150
K = np.linspace(K_min, K_max, 100) # Variable Strike Price

fig, ax = plt.subplots(figsize=(10, 4))

# 1. Draw the Barrier (The Spot Price)
ax.axvline(S, color='black', linestyle='--', lw=2)
ax.text(S, -0.1, f'Spot Price (S={S})', ha='center', fontweight='bold')

# 2. Represent the CALL Regions (Inverted logic)
# Left side: K < S (Strike below Spot) -> Call is ITM, Put is OTM
ax.axvspan(K_min, S, color='green', alpha=0.1, label='Call ITM / Put OTM')

# Right side: K > S (Strike above Spot) -> Call is OTM, Put is ITM
ax.axvspan(S, K_max, color='red', alpha=0.1, label='Call OTM / Put ITM')

# 3. Add Labels for the Mind
ax.text((K_min + S)/2, 0.5, "CALL: ITM\n(In the Money)", ha='center', color='darkgreen', fontsize=12)
ax.text((K_min + S)/2, 0.2, "PUT: OTM\n(Worthless)", ha='center', color='darkred', fontsize=10)

ax.text((K_max + S)/2, 0.5, "CALL: OTM\n(Worthless)", ha='center', color='darkred', fontsize=12)
ax.text((K_max + S)/2, 0.2, "PUT: ITM\n(In the Money)", ha='center', color='darkgreen', fontsize=10)

# Formatting
ax.set_xlim(K_min, K_max)
ax.set_ylim(0, 1)
ax.set_yticks([])
ax.set_xlabel("Strike Price (K)", fontsize=12)
ax.set_title("Moneyness Zones (Variable Strike)", fontsize=14)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```
