### At-The-Money (ATM) Call Price Approximation

When a call option is exactly At-The-Money ($S = K$), its intrinsic value is zero. The entire price consists of **extrinsic (time) value**, which can be approximated using a simplified version of the Black-Scholes model.

#### 1. The Approximation Formula
Under the assumption of **zero interest rates ($r=0$)**, the price of an ATM Call is:

$$C \approx 0.4 \cdot S \cdot \sigma\sqrt{T}$$

#### 2. Required Conditions
* **At-The-Money (ATM):** The current spot price must equal the strike price ($S = K$).
* **Zero Rates ($r = 0$):** The risk-free rate is assumed to be zero (or negligible).
* **Constant Volatility:** Assumes the Black-Scholes framework of constant implied volatility.

#### 3. Parameter Definitions
* **$S$**: Current Spot Price.
* **$\sigma$**: Implied Volatility (expressed as a decimal, e.g., 0.20 for 20%).
* **$T$**: Time to maturity in years.
* **$0.4$**: A rounding of the constant $\frac{1}{\sqrt{2\pi}} \approx 0.3989$.

---

#### Why it Works
Mathematically, at $S=K$ and $r=0$, the Black-Scholes formula simplifies through a first-order Taylor expansion. Since the call's value at the strike is essentially its **Vega** multiplied by its **volatility**, the price scales linearly with $\sigma$ and with the square root of time ($\sqrt{T}$).

**Example:**
For a stock at **$100**, with **20% vol** and **1 year** to expiry:
$$C \approx 0.4 \times 100 \times 0.20 \times \sqrt{1} = \$8.00$$
*(The exact Black-Scholes price is ~$7.98)*.
